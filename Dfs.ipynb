### Problem 3: Checking Routes Between Cities Using DFS

This problem involves using Depth First Search (DFS) to analyze a road network. We'll implement functions to:
1. Check if a route exists between two given cities.
2. Find all separate groups of connected cities (connected components).

```python
# 1. Represent the road network as an adjacency list
def create_graph(connections):
    graph = {}
    # Initialize graph with all cities to handle isolated nodes
    all_cities = set()
    for city1, city2 in connections:
        all_cities.add(city1)
        all_cities.add(city2)

    for city in all_cities:
        graph[city] = []

    for city1, city2 in connections:
        graph[city1].append(city2)
        graph[city2].append(city1) # Assuming roads are bidirectional
    return graph

# Depth First Search (DFS) function
def dfs(graph, start_node, visited):
    visited.add(start_node)
    for neighbor in graph.get(start_node, []):
        if neighbor not in visited:
            dfs(graph, neighbor, visited)
```

#### Part 1: Check whether a route exists between two given cities.

```python
def check_route(graph, start_city, end_city):
    visited = set()
    # Ensure both start and end cities exist in the graph
    if start_city not in graph or end_city not in graph:
        return False
    dfs(graph, start_city, visited)
    return end_city in visited

# Example Usage for Part 1
connections = [
    ('A', 'B'), ('B', 'C'), ('C', 'D'), ('D', 'A'), # Group 1
    ('E', 'F'), ('F', 'G'), # Group 2
    ('H', 'I') # Group 3
]
graph = create_graph(connections)

print(f"Route from A to C exists: {check_route(graph, 'A', 'C')}")
print(f"Route from A to F exists: {check_route(graph, 'A', 'F')}")
print(f"Route from E to G exists: {check_route(graph, 'E', 'G')}")
print(f"Route from H to J exists (J doesn't exist): {check_route(graph, 'H', 'J')}")
print(f"Route from H to I exists: {check_route(graph, 'H', 'I')}")
```

#### Part 2: Find all separate groups of cities (connected components).

```python
def find_connected_components(graph):
    visited = set()
    components = []
    for city in graph:
        if city not in visited:
            current_component = []
            component_visited_temp = set() # Use a temporary set for DFS to collect component nodes
            dfs(graph, city, component_visited_temp)

            # Add newly visited nodes to the overall visited set and the current component
            for node in component_visited_temp:
                if node not in visited:
                    visited.add(node)
                    current_component.append(node)

            if current_component: # Only add if the component is not empty
                components.append(current_component)
    return components

# Example Usage for Part 2
connections = [
    ('A', 'B'), ('B', 'C'), ('C', 'D'), ('D', 'A'),
    ('E', 'F'), ('F', 'G'),
    ('H', 'I')
]
graph = create_graph(connections)

print("\nConnected Components (Example 1):")
for i, component in enumerate(find_connected_components(graph)):
    print(f"Component {i+1}: {component}")

# Example with a disconnected node
all_cities_example_2 = {'X', 'Y', 'Z', 'P', 'Q', 'R'} # R is an isolated node
connections_example_2 = [
    ('X', 'Y'), ('Y', 'Z'),
    ('P', 'Q')
]

# Create graph ensuring all_cities_example_2 are part of the graph even if isolated
graph_example_2 = {city: [] for city in all_cities_example_2}
for city1, city2 in connections_example_2:
    graph_example_2[city1].append(city2)
    graph_example_2[city2].append(city1)

print("\nConnected Components (Example 2 with isolated node 'R'):")
for i, component in enumerate(find_connected_components(graph_example_2)):
    print(f"Component {i+1}: {component}")
```
