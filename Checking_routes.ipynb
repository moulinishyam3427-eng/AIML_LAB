```python
import matplotlib.pyplot as plt
import networkx as nx

# Create an undirected graph
G = nx.Graph()

# Add the road connections (edges) between cities
edges = [
    ("Pune", "Mumbai"),
    ("Bankura", "Durgapur"),
    ("Durgapur", "Kolkata"),
    ("Kolkata", "Asansol"),
    ("Asansol", "Dhanbad"),
    ("Dhanbad", "Ranchi"),
]
G.add_edges_from(edges)

# Define approximate positions to match the layout in the image
pos = {
    "Pune": (1, 4),
    "Mumbai": (3, 4.5),
    "Bankura": (0, 1.5),
    "Durgapur": (2, 1.7),
    "Kolkata": (2, 0),
    "Asansol": (4, 0),
    "Dhanbad": (5.5, 1.2),
    "Ranchi": (5, 0.5),
}

# Draw the network layout
plt.figure(figsize=(8, 5))
nx.draw_networkx_nodes(
    G, pos, node_color="#e0b88a", node_size=1800, edgecolors="none"
)
nx.draw_networkx_edges(G, pos, edge_color="gray", width=2)
nx.draw_networkx_labels(G, pos, font_size=10, font_weight="bold")

# Customize view
plt.title("City Road Network", fontsize=14, fontweight="bold")
plt.axis("off")
plt.tight_layout()
plt.show()
```
